```python
import csv
import random
```

# 1. Create Image File

```python
def create_image_file(filename:str, size:int) -> list[list[int]]:
    '''
    Creates a grayscale image file containing random data with user-defined size.
    '''
    image_wdata = []
    for x in range(0, size):
        row = []
        for y in range(0, size):
            row.append(random.randint(0, 255))   
        image_wdata.append(row)
    f1 = open(filename, 'w')
    c1 = csv.writer(f1)
    c1.writerow([z for z in range(1, size + 1)])
    c1.writerows(image_wdata)
    f1.close()
    return image_wdata
```

```python
create_image_file('image_file.csv', 10)  
```

# 2. Load grayscale image from text file

```python
def load_image(filename:str) -> list[list[int]]:
    '''
    Loads an image from a text file and returns a 2D list.
    '''
    f1 = open(filename, 'r')
    c1 = csv.reader(f1)
    image_rdata = []
    next(c1)
    for row in c1:
        if row != []:      # Reads the row if it is not an empty row.
            image_rdata.append(row)
    f1.close()
    return image_rdata
```

```python
image_rdata = load_image('image_file.csv')
```

```python
image_rdata
```

# 3. Convert Image Data to Integer

```python
def convert_to_int(image_data: list[list[str]]) -> list[list[int]]:
    '''
    Converts image pixels which are read as strings to integers.
    '''
    image_rdata_int = []
    for row in image_data:
        list1 = []
        for x in row:
            x = int(x)    
            list1.append(x)
        image_rdata_int.append(list1)
    return image_rdata_int
```

```python
image_rdata_int = convert_to_int(image_rdata)
```

```python
image_rdata_int
```

# 4. Invert Image

```python
def invert_image(image_data: list[list[int]]) -> list[list[int]]:
    '''
    Inverts grayscale image.
    new_pixel = 255 - pixel
    '''
    image_rdata_invert = []
    for row in image_data:
        list1 = []
        for x in row:
            x = 255 - x
            list1.append(x)
        image_rdata_invert.append(list1)
    return image_rdata_invert
```

```python
image_rdata_invert = invert_image(image_rdata_int)
```

# 5. Adjust Brightness

```python
def adjust_brightness(image_data: list[list[int]], value: int) -> list[list[int]]:
    '''
    Adjusts brightness by adding value to each pixel.
    '''
    image_rdata_brightness = []
    for row in image_data:
        list1 = []
        for x in row:
            if x + value <= 255 and x + value >= 0:
                x = x + value
            elif x + value > 255:   # Clamps to 255
                x = 255
            elif x + value < 0:     # Clamps to 0
                x=0
            list1.append(x)
        image_rdata_brightness.append(list1)
    return image_rdata_brightness
```

```python
image_rdata_brightness = adjust_brightness(image_rdata_int, 20)
```

# 6. Threshold Filter

```python
def threshold_filter(image_data: list[list[int]], threshold: int) -> list[list[int]]:
    '''
    Converts image to black and white.
    '''
    image_rdata_threshold = []
    for row in image_data:
        list1 = []
        for x in row:
            if x <= threshold :
                x = 0
            else:
                x = 255
            list1.append(x)
        image_rdata_threshold.append(list1)
    return image_rdata_threshold
```

```python
image_rdata_threshold = threshold_filter(image_rdata_int, 100)
```

# 7. Horizontal Edge Detection

```python
def detect_edges(image_data: list[list[int]]) -> list[list[int]]:
    '''
    Horizontal edge detection.
    '''
    image_rdata_edge = []
    for row in image_data:
        list1 = []
        list1.append(0)       #The first element of the row does not have a left neighbour, so we consider the edge for that index as zero.
        for x in range(1, len(row)): 
            list1.append(row[x] - row[x - 1])
        image_rdata_edge.append(list1)
    return image_rdata_edge
```

```python
image_rdata_edge = detect_edges(image_rdata_int)
```

# 8. Save Image

```python
def save_image(image_data: list[list[int]], filename: str) -> None:
    '''
    Saves image to a file
    '''
    f1 = open(filename,'w')
    c1 = csv.writer(f1)
    c1.writerow([x for x in range(1, len(image_data) + 1)])
    c1.writerows(image_data)
    f1.close()
```

```python
save_image(image_rdata_edge, 'image_file.csv')
```

# 9. Menu

```python
def menu():
    while True:
        try:
            user_choice_1 = int(input('Select number 1 to create an image file:'))
            if user_choice_1 == 1:       
                user_file_name = input('Enter filename:')   
            else:
                print('Wrong number!')
                continue
            user_size = int(input('Select image size:'))
            break
        except:  
            print('Wrong number!')
            
    user_image_wdata = create_image_file(user_file_name, user_size)
    print(user_image_wdata)
    
    while True:
        try:
            user_choice_2 = int(input('Select number 2 to load the image from the text file:'))
            if user_choice_2 != 2:
                continue
            else:
                break
        except:
            print('Wrong number!')
            
    user_image_rdata = load_image(user_file_name)
    print(user_image_rdata)
    
    while True:
        try:
            user_choice_3 = int(input('Select number 3 to convert the image data to integer:'))
            if user_choice_3 != 3:
                continue
            else:
                break
        except:
            print('Wrong number!')
            
    user_image_rdata_int = convert_to_int(user_image_rdata)
    print(user_image_rdata_int)
    user_image_rdata_current = user_image_rdata_int
    
    while True:
        try:
            user_choice = int(input('''Select a function number:
            4. Invert image
            5. Adjust brightness
            6. Threshold filter
            7. Horizontal edge detection
            8. Save image:
            '''))
        except:
            print('Wrong number!')
            continue
        if user_choice == 4:
            user_image_rdata_current = invert_image(user_image_rdata_current)
            print(user_image_rdata_current)
        elif user_choice == 5:
            while True:
                try:
                    user_brightness_value = int(input('Enter the brightness value:'))
                    if user_brightness_value >= 0 and user_brightness_value <= 255:
                        user_image_rdata_current = adjust_brightness(user_image_rdata_current, user_brightness_value)
                        print(user_image_rdata_current)
                        break
                    else:
                        print('Out of range. Enter a number between 0 and 255')
                except:
                    print('Wrong value!')
        elif user_choice == 6:
            while True:
                try:
                    user_threshold_value = int(input('Enter the threshold value:'))
                    if user_threshold_value >= 0 and user_threshold_value <= 255:
                        user_image_rdata_current = threshold_filter(user_image_rdata_current, user_threshold_value)
                        print(user_image_rdata_current)
                        break
                    else:
                        print('Out of range. Enter a number between 0 and 255')
                except:
                    print('Wrong value!')
        elif user_choice == 7:
            user_image_rdata_current = detect_edges(user_image_rdata_current)
            print(user_image_rdata_current)
        elif user_choice == 8:
            new_user_file_name = input('Enter filename:')
            save_image(user_image_rdata_current, new_user_file_name)
            break
        else:
            print('Wrong number!')
        
        
        
```
